```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

# ── Confusion matrix values ──────────────────────────────────────────────────
# Rows = Actual class  (Healthy, Sick)
# Cols = Predicted class (Healthy, Sick)
cm = np.array([
    [14, 1],   # Actual Healthy → TN=14, FP=1
    [2,  13]   # Actual Sick    → FN=2,  TP=13
])

labels = ['Healthy', 'Sick']

# ── Cell annotations with label ──────────────────────────────────────────────
annot = np.array([
    ['TN = 14', 'FP = 1'],
    ['FN = 2',  'TP = 13']
])

# ── Colour map: correct (diagonal) vs incorrect (off-diagonal) ───────────────
cell_colors = np.array([
    ['#2ecc71', '#e74c3c'],   # TN=green, FP=red
    ['#e74c3c', '#2ecc71']    # FN=red,   TP=green
])

fig, ax = plt.subplots(figsize=(7, 6))

# Draw coloured cells manually
for i in range(2):
    for j in range(2):
        rect = mpatches.FancyBboxPatch(
            (j, 1 - i), 1, 1,
            boxstyle='round,pad=0.05',
            linewidth=2,
            edgecolor='white',
            facecolor=cell_colors[i, j],
            alpha=0.85,
            transform=ax.transData,
            clip_on=False
        )
        ax.add_patch(rect)
        ax.text(
            j + 0.5, 1 - i + 0.5,
            annot[i, j],
            ha='center', va='center',
            fontsize=15, fontweight='bold', color='white'
        )

# ── Axes formatting ──────────────────────────────────────────────────────────
ax.set_xlim(0, 2)
ax.set_ylim(0, 2)
ax.set_xticks([0.5, 1.5])
ax.set_xticklabels(labels, fontsize=13)
ax.set_yticks([0.5, 1.5])
ax.set_yticklabels(['Sick', 'Healthy'], fontsize=13)
ax.xaxis.set_ticks_position('top')
ax.xaxis.set_label_position('top')
ax.set_xlabel('Predicted Label', fontsize=13, fontweight='bold', labelpad=10)
ax.set_ylabel('Actual Label', fontsize=13, fontweight='bold', labelpad=10)
ax.tick_params(length=0)
for spine in ax.spines.values():
    spine.set_visible(False)



# ── Legend ───────────────────────────────────────────────────────────────────
legend_handles = [
    mpatches.Patch(color='#2ecc71', label='Correct Prediction (TP / TN)'),
    mpatches.Patch(color='#e74c3c', label='Incorrect Prediction (FP / FN)')
]
ax.legend(handles=legend_handles, loc='lower center',
          bbox_to_anchor=(0.5, -0.1), fontsize=11, frameon=False)

plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
print('Figure saved as confusion_matrix.png')
```

## Performance Metrics

Using the confusion matrix values we can compute standard evaluation metrics:

```python
TP = 13
TN = 14
FP = 1
FN = 2
total = TP + TN + FP + FN

accuracy    = (TP + TN) / total
precision   = TP / (TP + FP)
recall      = TP / (TP + FN)          # Sensitivity
specificity = TN / (TN + FP)
f1          = 2 * (precision * recall) / (precision + recall)

print(f"{'Metric':<20} {'Value':>10}")
print("-" * 32)
print(f"{'Total Test Records':<20} {total:>10}")
print(f"{'Correct Predictions':<20} {TP + TN:>10}")
print(f"{'Wrong Predictions':<20} {FP + FN:>10}")
print("-" * 32)
print(f"{'Accuracy':<20} {accuracy:>9.1%}")
print(f"{'Precision':<20} {precision:>9.1%}")
print(f"{'Recall (Sensitivity)':<20} {recall:>9.1%}")
print(f"{'Specificity':<20} {specificity:>9.1%}")
print(f"{'F1 Score':<20} {f1:>9.4f}")
```

## Interpretation

| Metric | Value | Meaning |
|--------|-------|---------|
| **Accuracy** | 90% | The model correctly classified 27 out of 30 patients |
| **Precision** | 92.9% | When the model predicts Sick, it is correct ~93% of the time |
| **Recall** | 86.7% | The model detected ~87% of all actually Sick patients |
| **Specificity** | 93.3% | The model correctly identified ~93% of Healthy patients |
| **F1 Score** | 0.8966 | Balanced measure of Precision and Recall |

### Key Observation
> The **2 False Negatives** are the most critical errors in a healthcare context — sick patients sent home without treatment. Minimising FN should be prioritised, even at the cost of a slightly higher FP rate.
